Exploratory Data Analysis

```python
import pandas as pd
df = pd.read_csv("../data/raw/creditcard.csv")
# print(df.shape)
# df.head()
```

```python
# Sampling first
df = df.sample(n=50000)
```

1. Basic Inspection

```python
ttlRow = df.shape[0]
ttlCol = df.shape[1]
# print(ttlRow)
# print(ttlCol)

columnInfo = pd.DataFrame({
    "columnName": df.columns,
    "Data Type": df.dtypes
})
# print(columnInfo)

basicInfo = df.info()
```

2. Data Quality Check

```python
# checkIsNull = df.isnull().any().any()
checkIsNull = df.isnull().sum()
# print(checkIsNull)

# 1. 先检查缺失
print(df.isnull().sum())

# 2. 数值列用均值填充
df["Amount"] = df["Amount"].fillna(df["Amount"].mean())

# 3. 分类列用众数填充 （第一个众数）
# df["Class"] = df["Class"].fillna(df["Class"].mode()[0])

# 4. 剩余少量缺失行直接删除
df = df.dropna()
```

3. Statistical Analysis

```python
classUniq = df["Class"].unique()
# print(classUniq)

classDistri = df["Class"].value_counts()
print(classDistri)

classDisNor = df["Class"].value_counts(normalize=True)
print(classDisNor)
```

```python
amountDis = df["Amount"].value_counts()
print(amountDis)
```

4. Visual Analysis

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig_amount_hist = plt.figure(figsize=(10, 6))
plt.hist(df["Amount"], bins=10, edgecolor="black")

plt.title("Amount Frequency Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
fig_amount_by_class_box = plt.figure(figsize=(10,6))
sns.boxplot(x="Class", y="Amount", data=df)
plt.title("Amount Distribution Across Class")
plt.show()
```

```python
df.iloc[:1000, 0]

timeDis = df["Time"].value_counts()
print(timeDis)

fig_time_hist = plt.figure(figsize=(10, 6))
plt.hist(df["Time"], bins=10, edgecolor="black")

plt.title("Time Frequency Distribution")
plt.xlabel("Time")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

5. Correlation Analysis

```python
plt.figure(figsize=(10,8))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Features Correlation Heatmap")
plt.tight_layout()
plt.show()
```
